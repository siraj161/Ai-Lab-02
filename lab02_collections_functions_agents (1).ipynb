## 8. Vacuum Cleaner Agent
The vacuum agent observes its location and whether that location is dirty, then chooses an action.

### Examples

```python
# Example 1: One-room decision
def vacuum_agent(location, status):
    if status == "dirty":
        return "Suck"
    if location == "A":
        return "Right"
    return "Left"

print(vacuum_agent("A", "dirty"))
```

```python
# Example 2: Multiple percepts
percepts = [("A", "dirty"), ("A", "clean"), ("B", "dirty")]
for location, status in percepts:
    print(location, status, vacuum_agent(location, status))
```

```python
# Example 3: Track actions
actions = []
percepts = [("A", "dirty"), ("A", "clean"), ("B", "dirty")]
for location, status in percepts:
    actions.append(vacuum_agent(location, status))
print(actions)
```

### Practice Questions — Solutions

```python
# Q1. Run vacuum_agent for A dirty, A clean, B dirty and B clean.
def vacuum_agent(location, status):
    if location not in ("A", "B"):
        return "Unknown Location"
    if status == "dirty":
        return "Suck"
    if status != "clean":
        return "Unknown Status"
    if location == "A":
        return "Right"
    return "Left"

for percept in [("A", "dirty"), ("A", "clean"), ("B", "dirty"), ("B", "clean")]:
    print(percept, "->", vacuum_agent(*percept))
```

```python
# Q2. Store four vacuum percepts in a list of tuples and print the action for each.
percepts = [("A", "dirty"), ("A", "clean"), ("B", "dirty"), ("B", "clean")]
for location, status in percepts:
    print(location, status, "->", vacuum_agent(location, status))
```

```python
# Q3. Unknown location returns Unknown Location.
print(vacuum_agent("C", "clean"))
```

```python
# Q4. Store all chosen actions in a list and print the final list.
percepts = [("A", "dirty"), ("A", "clean"), ("B", "dirty"), ("B", "clean")]
actions = []
for location, status in percepts:
    actions.append(vacuum_agent(location, status))
print(actions)
```

```python
# Q5. Count how many times the action Suck occurs.
suck_count = actions.count("Suck")
print("Suck actions:", suck_count)
```

```python
# Q6. Create a visited set containing every location observed.
visited = set()
for location, status in percepts:
    visited.add(location)
print(visited)
```

```python
# Q7. Dictionary holding location, status and chosen_action for one percept.
location, status = ("A", "dirty")
percept_record = {
    "location": location,
    "status": status,
    "chosen_action": vacuum_agent(location, status)
}
print(percept_record)
```

**Q8. Why is this a simple reflex agent?**

It is a simple reflex agent because it selects an action directly from the current percept using condition-based rules, without planning from a history of past percepts.

### Challenge 5: Function choose_cheapest(costs)

```python
def choose_cheapest(costs):
    if not costs:
        return None
    cheapest = costs[0]
    for cost in costs:
        if cost < cheapest:
            cheapest = cost
    return cheapest

print(choose_cheapest([15, 8, 21, 4, 10]))
```
